```python
from pathlib import Path

# 1. 创建 data 文件夹
data_dir = Path("data")
data_dir.mkdir(exist_ok=True)

# 2. 拼接文件路径
file_path = data_dir / "example.txt"

# 3. 创建文件并写入内容
file_path.write_text(
    "这是我的第一个 RAG 项目",
    encoding="utf-8"
)

# 4. 判断文件是否存在
print(file_path.exists())

# 5. 读取文件
text = file_path.read_text(encoding="utf-8")
print(text)
```
